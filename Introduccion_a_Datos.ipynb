```python
# Escribir archivo TXT de ejemplo
with open('datos.txt', 'w') as f:
    f.write("""Hola
    Mi
    NOMBRE
    Es
    DANIEL PIÑOL
    Y
    ME GUSTA EL FUTBOL""")

# Escribir archivo CSV de ejemplo
import csv
with open('datos.csv', 'w', newline='') as f:
    escritor = csv.writer(f)
    escritor.writerow(['Nombre', 'Edad', 'Ciudad'])
    escritor.writerow(['JUAN', 30, 'Guatemala'])
    escritor.writerow(['XIMENA', 29, 'Antigua'])
    escritor.writerow(['SERGIO', 19, 'Petén'])

print("Archivos de ejemplo creados")
```

```python
# Leer archivo TXT usando context manager
with open('datos.txt', 'r') as f:
    contenido = f.read()
    print("Contenido del archivo TXT:")
    print(contenido)
```

```python
import csv

# Leer archivo CSV usando context manager
with open('datos.csv', 'r') as f:
    lector = csv.reader(f)
    print("Contenido del archivo CSV:")
    for fila in lector:
        print(fila)
```

¿En qué parte o partes de un Pipeline de Datos se pueden usar las operaciones de Lectura (R) y Escritura (W)?

Se usan en todas las etapas. Lectura para traer datos desde archivos/bases de datos, y escritura para guardar los resultados procesados. Están presentes desde que se obtienen los datos hasta que se almacenan en su destino final.

## Ejercicio 3: Archivos Pickle

### 3.1 Escritura de archivo Pickle

```python
import pickle

# Crear diferentes tipos de objetos para guardar
datos_dict = {
    'nombre': 'Análisis de Datos',
    'valores': [10, 20, 30, 40, 50],
    'metadata': {'fecha': '2026-04-10', 'versión': 1.0}
}

datos_lista = ['Python', 'Pandas', 'NumPy', 'Pickle']

# Guardar diccionario en Pickle
with open('pickle.pkl', 'wb') as f:
    pickle.dump(datos_dict, f)
    print("Diccionario guardado en Pickle")

# Guardar lista en Pickle
with open('lista_pikle.pkl', 'wb') as f:
    pickle.dump(datos_lista, f)
    print("Lista guardada en Pickle")
```

```python
# Cargar diccionario desde Pickle
with open('datos.pkl', 'rb') as f:
    datos_cargados = pickle.load(f)
    print("Datos cargados desde Pickle:")
    print(datos_cargados)
    print(f"Tipo de dato: {type(datos_cargados)}")

# Cargar lista desde Pickle
with open('lista.pkl', 'rb') as f:
    lista_cargada = pickle.load(f)
    print("\nLista cargada desde Pickle:")
    print(lista_cargada)
    print(f"Tipo de dato: {type(lista_cargada)}")
```

```python
# Crear un DataFrame
df_productos = pd.DataFrame({
    'Producto': ['Laptop', 'Mouse', 'Teclado'],
    'Precio': [15000, 250, 800],
    'Stock': [10, 50, 30]
})

print("DataFrame original:")
print(df_productos)

# Guardar DataFrame en Pickle
with open('dataframe.pkl', 'wb') as f:
    pickle.dump(df_productos, f)
    print("\n✅ DataFrame guardado en Pickle")

# Cargar DataFrame desde Pickle
with open('dataframe.pkl', 'rb') as f:
    df_cargado = pickle.load(f)
    print("\nDataFrame cargado desde Pickle:")
    print(df_cargado)
```

### Pregunta 3: ¿Qué tipo de dato se podría almacenar con un Pickle?

Se puede almacenar casi cualquier objeto de Python: números, texto, listas, diccionarios, DataFrames, arrays de NumPy y modelos de Machine Learning. Se usa principalmente para guardar modelos entrenados (evita reentrenar), DataFrames procesados y resultados de análisis complejos. Solo funciona en Python y no debe cargarse de fuentes no confiables por seguridad.
